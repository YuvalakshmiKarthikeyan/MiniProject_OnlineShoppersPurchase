This project leverages data analytics and machine learning techniques to predict the purchasing intentions of online shoppers. 1.2 Project Objectives

</div>



<p style="

font-family:Arial;
">
The primary objectives of this project are:

* To analyze the online shoppers' browsing behavior and identify the factors that influence purchase intention.
* To perform Exploratory Data Analysis (EDA) to understand the dataset and discover meaningful patterns and relationships among the features.
* To preprocess the dataset by handling missing values, encoding categorical variables, treating outliers, and scaling numerical features where required.
* To build and compare multiple supervised machine learning classification models for predicting whether a visitor will complete a purchase.
* To evaluate the performance of each model using classification metrics such as Accuracy, Precision, Recall, F1-Score, and ROC-AUC.
* To identify the most important features that contribute to customer purchase decisions.
* To select the best-performing model based on evaluation results and provide business insights that can help e-commerce companies improve customer engagement and increase conversion rates.
</p>

4.3 Outlier Detection & Treatment</div></h3>

```python
outlier_summary = {}
outlier_cols = [
    'Administrative',
    'Administrative_Duration',
    'Informational',
    'Informational_Duration',
    'ProductRelated',
    'ProductRelated_Duration',
    'BounceRates',
    'ExitRates',
    'PageValues',
    'SpecialDay'
]
for col in outlier_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    outlier_summary[col] = len(outliers)

# Display outlier count
outlier_summary = pd.DataFrame(list(outlier_summary.items()),columns=['Feature', 'Outlier_Count'])
outlier_summary
```

Instead of removing data, extreme values can be capped within acceptable limits:

```python
# Create a copy of the original dataset
df_outlier_treated = df.copy()

for col in outlier_cols:

    Q1 = df_outlier_treated[col].quantile(0.25)
    Q3 = df_outlier_treated[col].quantile(0.75)
    IQR = Q3 - Q1

    # Skip columns with IQR = 0
    if IQR == 0:
        print(f"Skipping {col} (IQR = 0)")
        continue

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    df_outlier_treated[col] = df_outlier_treated[col].clip(
        lower=lower_bound,
        upper=upper_bound
    )

print("Outlier treatment completed.")
```
